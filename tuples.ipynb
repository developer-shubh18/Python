## Introduction to tuples

tuples are ordered collections of items that are immutable. They r similar to lists, but their immutability makes them different.


```python
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tp1=tuple()
print(type(tp1))
```

```python
numbers=tuple([1,2,3,4,5,6,7])
numbers
```

```python
list((1,2,3,4,5,6,7,8,9))
```

```python
mixed_tuple=(1,"hello",6.66,"shubh")
mixed_tuple
```

```python
## imutable nature of tuples
## tuple are imutable, meaning their value cannot be changed once assigned 

list=[1,2,4,5,6,8]
print(list)

list[1]="shubh"
print(list)
```

```python
##tuple methods 
print(numbers.count(1))
print(numbers.index(3))
```

```python
## packing and unpacking tuples
packed_tuples=1,"hello",3.14
print(packed_tuples)
```

```python
##unpacking tuple
a,b,c=packed_tuples
print(a)
print(b)
print(c)
```

```python
## unpacking with *
numbers=[1,2,3,4,5,6,7,8]
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuple
list=[[1,2,3,4],[3,4.17,4,5],["hello",6,7,"c"]]
list[2][0]
```

```python
list=[[1,2,3,4],[3,4.17,4,5],("hello",6,7,"c")]
list[2][0:3]
```

```python
nested_tuple=((1,2,3),("a","b","c"),(True,False))
print(nested_tuple[0])
print(nested_tuple[1][2])
```
