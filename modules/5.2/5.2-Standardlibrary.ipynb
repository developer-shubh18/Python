### Generating Random Numbers
The random module is commonly used for generating random numbers. Two important functions are randint, which returns a random integer within a specified range, and choice, which selects a random element from a collection.

```python
import random
print(random.randint(1, 10))
fruits = ['apple', 'banana', 'cherry']
print(random.choice(fruits))
```

### Data Serialization with JSON
Data serialization, such as converting a dictionary into JSON, is important when working with APIs or databases. The json module provides dumps to convert a dictionary to a JSON string and loads to convert a JSON string back to a dictionary.



```python
import json
data = {'name': 'Krish', 'age': 25}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

### Working with CSV Files
The csv module is used for reading and writing CSV files. You can write rows to a CSV file and then read them back using the appropriate reader and writer functions.



```python
import csv
with open('example.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Krish', 32])
with open('example.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# Date and Time Operations
from datetime import datetime, timedelta
now = datetime.now()
print(now)
yesterday = now - timedelta(days=1)
print(yesterday)
```

```python
# Program Timing with Time Module
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
# Regular Expressions with re Module
import re
pattern = r'\d+'
text = 'There are 123 apples.'
match = re.search(pattern, text)
if match:
    print(match.group())
```
