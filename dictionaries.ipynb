## Dictionaries
## introduction to Dictionaries
dict are unordered collection of items. they store data in key values pairs key must be unique and immutable while values can be of any type

```python
## creating a dict
empty_dict={}
print(type(empty_dict))
```

```python
students={"name":"shubh","age":15,"grade":"A"}
print(students)
print(type(students))
#if u redefine a key the value it access is the last updated value
```

```python
students={"name":"shubh","age":15,"grade":"A"}
print(students)
```

```python
# accessing dict elements
print(students['grade'])
print(students['age'])
#accessing through get methods 
print(students.get('grade'))
print(students.get('last_name'))
print(students.get('last_name',"kumar"))
```

```python
## Modifying dictionaries element
## dict are mutable so u can add updATE OR DELETE

print(students)
```

```python
students['age']=33 ##update
print(students)
students["address"]="BRC"## add
print(students)
del students['grade'] ##delete
```

```python
##dictionary methods

keys=students.keys()
print(keys)
values=students.values()
print(values)
items=students.items()
print(items)
```

```python
## shallow copy

student_copy=students
print(students)
print(student_copy)
```

```python
students["name"]="shubh2"
print(students)
print(student_copy)
```

```python
student_copy1=students.copy()
print(student_copy1)
print(students)
```

```python
students["name"]="shubh3"
print(student_copy1)
print(students)
```

```python
# iterating over dict
# you canuse loops to iterate over dict keys values or items 
# iterating over keys

for keys in students.keys():
    print(keys)
```

```python
# iterate over values
for values in students.values():
    print(values)
#iterate over items 
for items in students.items():
    print(items)

#iterate over key value pairs
for key,value in students.items():
    print(f"{key}:{value}")
```

```python
## NESTED DICTONARIES
student={
    "student1":{"name":"shubh","age":18,"address":"brc"},
    "student2":{"name":"virat","age":38,"address":"mum"},
    "student3":{"name":"tony","age":28,"address":"del"}
}
print(student)
```

```python
## Access nested dic ele
print(student["student2"]["name"])
print(student["student2"])
```

```python
student.items()
```

```python
## Iterating over nested dictionaries
for student_id, student_info in student.items():
    print(f"(student_id):{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
## dictionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Example
# use a dic to count the freq of elements in list
numberss=[1,1,1,2,3,3,4,4,5,3,5,5,6,6,7,7,7,8]
frequency={}
for number in numberss:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## merge 2 dictionaries

dict1={"a":1,"b":2}
dict2={"c":3,"d":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
