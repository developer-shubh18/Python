## Sets
sets are a built in data type in python used to store collections of unique items. they are unordered meaning that the elements do not follow a specific order, and they donot allow duplicate elements. Sets are useful for membership tests, eliminating duplicate enteries, and performing mathematical set operation like union, intersection, difference, and symmetric difference.



```python
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6,7])
print((my_set))
```

```python
## Basics Sets operation
## Adding and removing  Elements
my_set.add(7)
print(my_set)
my_set.add(18)
print(my_set)
```

```python
my_set.remove(3)
print(my_set)
```

```python
my_set.discard(10)##discrd will check for the element whether its present or not.
print(my_set)
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
my_set.clear()
print(my_set)
```

```python
##Set Membership Test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical Operation
set1={1,2,3,4,5,6}
set2={1,20,33,5,6,7,8}
union_set=set1.union(set2)
print(union_set)
intersection_set=set1.intersection(set2)
print(intersection_set)
set1.intersection_update(set2)
print(set1)
```

```python
set2.difference(set1)
```

```python
set2.symmetric_difference(set1)
```

```python
## set methods
set1={1,2,3,4}
set2={5,6,7,8}
## is subset
print(set1.issubset(set2))
print(set1.issuperset(set2))
```
