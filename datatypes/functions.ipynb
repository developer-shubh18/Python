### What is a Function?
A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability. It is not necessary for a function to always return a value, but many functions do.

```python
#syntax
def my_function():
    """This is a docstring."""
    # function body
    return
```

### Why Do We Require Functions?
Suppose you need to check whether a number is even or odd in multiple places in your project. Instead of copying the same code everywhere, you can write it once inside a function and reuse it.

```python
num = 24
if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def even_or_odd(num):
    if num % 2 == 0:
        """"This function finds even or odd"""
        print("The number is even")
    else:
        print("The number is odd")
```

```python
even_or_odd(24)
```

### Default Parameters
You can provide default values to function parameters. If the argument is not provided during the function call, the default value is used.

```python
def greet(name="guest"):
    print(f"Hello {name}. Welcome to the Paradise.")
```

```python
greet("Krish")
greet()
```

### Variable Length Arguments
Functions can accept a variable number of arguments using *args for positional arguments and **kwargs for keyword arguments

```python
def print_numbers(*krish):
    for number in krish:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

## keywords Arguments
Keyword arguments are provided as key-value pairs and are accessed using **kwargs.

```python
# Keyword args
def print_detail(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_detail(name="Krish", age=32, country="India")
```

### Combining Positional and Keyword Arguments
You can combine both *args and **kwargs in a single function. Positional arguments must come before keyword arguments.

```python
def print_detail(*args, **kwargs):
    print("Positional arguments:")
    for val in args:
        print(val)
    print("Keyword arguments:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_detail(1, 2, 3, 4, 5, name="Krish", age=32)
```

## The Return Statement
The return statement is used to return values from a function. Functions can return single or multiple values.

```python
def multiply(a, b):
    return a * b
```

```python
result = multiply(2, 3)
print(result)
```

### Returning Multiple Values
A function in Python can also return multiple values, which are returned as a tuple.

```python
def multi_return(a, b):
    return a, b, a * b
```

```python
x, y, prod = multi_return(2, 3)
print(x, y, prod)
```
